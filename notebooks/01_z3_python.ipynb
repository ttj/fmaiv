## 1. Solve a system of constraints

Find integers `x, y` with `x + y == 10` and `x - y == 2`.

```python
from z3 import *

x, y = Ints('x y')
s = Solver()
s.add(x + y == 10, x - y == 2)
print(s.check())     # sat
print(s.model())     # [x = 6, y = 4]
```

## 2. Prove a property (check the negation is `unsat`)

To *prove* `x*x >= 0` for all reals `x`, ask Z3 whether the **negation** is
satisfiable. `unsat` means no counterexample exists, so the property holds.

```python
x = Real('x')
s = Solver()
s.add(Not(x * x >= 0))   # negation of the property
print(s.check())         # unsat  =>  property holds
```

## 3. Bit-vectors

Solve `x ^ 0x2A == 0xFF` over 8-bit machine integers.

```python
x = BitVec('x', 8)
s = Solver()
s.add(x ^ 0x2A == 0xFF)
print(s.check())
print(hex(s.model()[x].as_long()))
```
